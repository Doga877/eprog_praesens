## Teil 1: Die Klasse `Datensatz`

Unsere erste Klasse verwaltet die Daten. Sie hält zwei Arrays:

- `X` die Eigenschaften, eine Zeile pro Molekül, eine Spalte pro Eigenschaft
- `y` die Antwort, also 0 oder 1

Diese Benennung ist in Machine Learning überall üblich.

| Spalte | Bedeutung |
|---|---|
| `MW` | Molekülmasse |
| `AlogP` | Fettlöslichkeit |
| `HBA` | Wasserstoffbrücken-Akzeptoren |
| `HBD` | Wasserstoffbrücken-Donoren |
| `RB` | drehbare Bindungen |
| `Class` | **die Antwort**: 1 = wirkt, 0 = wirkt nicht |

```python
class Datensatz:
    """Hält die Molekül-Eigenschaften (X) und die Antwort (y)."""

    def __init__(self, dateiname):
        # names=True nimmt die erste Zeile der CSV als Spaltennamen
        tabelle = np.genfromtxt(dateiname, delimiter=",", names=True)

        self.namen = ["MW", "AlogP", "HBA", "HBD", "RB"]
        self.X = np.column_stack([tabelle[name] for name in self.namen])
        self.y = tabelle["Class"]

    def anzahl(self):
        return len(self.y)

    def anzahl_aktiv(self):
        return int(self.y.sum())      # die Nullen tragen nichts bei

    def nur_aktive(self):
        """Boolean Masking: nur die Zeilen, bei denen y gleich 1 ist."""
        return self.X[self.y == 1]

    def nur_inaktive(self):
        return self.X[self.y == 0]

    def vergleich_tabelle(self):
        """Zeigt pro Eigenschaft den Mittelwert beider Gruppen."""
        print(f"{'Eigenschaft':<12}{'aktiv':>10}{'inaktiv':>10}")
        print("-" * 32)

        for i, name in enumerate(self.namen):
            # axis=0 heisst: spaltenweise rechnen
            print(f"{name:<12}"
                  f"{self.nur_aktive()[:, i].mean():>10.2f}"
                  f"{self.nur_inaktive()[:, i].mean():>10.2f}")

    def plot(self, name="MW"):
        i = self.namen.index(name)
        plt.hist(self.nur_aktive()[:, i], bins=30, alpha=0.6, label="aktiv")
        plt.hist(self.nur_inaktive()[:, i], bins=30, alpha=0.6, label="inaktiv")
        plt.xlabel(name)
        plt.ylabel("Anzahl Moleküle")
        plt.legend()
        plt.show()
```

Jetzt erzeugen wir ein Objekt dieser Klasse und benutzen seine Methoden.

```python
daten = Datensatz("bace.csv")

print("Moleküle insgesamt:", daten.anzahl())
print("Davon aktiv:       ", daten.anzahl_aktiv())
print("Anteil aktiv:      ", f"{daten.y.mean():.1%}")
print()
print("Form von X:", daten.X.shape)      # (Zeilen, Spalten)
print("Die ersten drei Moleküle:")
print(daten.X[:3])
```

### Wie sieht Boolean Masking von innen aus?

`daten.y == 1` ergibt ein Array aus `True` und `False`. Setzt man das in die
eckigen Klammern von `X`, bekommt man nur die passenden Zeilen. Mit Listen
ginge das nicht.

```python
maske = (daten.y == 1)
print("Die ersten zehn Einträge der Maske:", maske[:10])

print()
print("Form von nur_aktive():  ", daten.nur_aktive().shape)
print("Form von nur_inaktive():", daten.nur_inaktive().shape)
```

### Unterscheiden sich die beiden Gruppen überhaupt?

```python
daten.vergleich_tabelle()
```

Die aktiven Moleküle sind im Schnitt schwerer und haben mehr von fast allem.
Reicht das zum Vorhersagen? Schauen wir es uns an.

```python
daten.plot("MW")
```

Die Verteilungen überlappen stark. Keine einzelne Eigenschaft entscheidet die
Frage, aber vielleicht gelingt es mit allen fünf zusammen. Genau dafür nehmen
wir Machine Learning.

## Teil 2: Die Klasse `Experiment`

Die zweite Klasse führt den Ablauf durch. Sie hat für jeden Schritt eine
Methode, und `run()` ruft sie der Reihe nach auf.

| Methode | Aufgabe |
|---|---|
| `aufteilen()` | Daten in Training und Test trennen |
| `skalieren()` | Spalten vergleichbar machen |
| `trainieren()` | das Netz lernen lassen |
| `auswerten()` | Genauigkeit und Verlust ausgeben |
| `vorhersagen()` | ein einzelnes Molekül beurteilen |
| `vergleiche()` | mehrere Einstellungen ausprobieren |

Die Einstellungen des Netzes dürft ihr selbst wählen:

| Einstellung | Bedeutung | Mögliche Werte |
|---|---|---|
| `neuronen` | Schichten und Neuronen | `(8,)`, `(32, 16)`, `(64, 32, 16)` |
| `aktivierung` | Rechenregel pro Neuron | `"relu"`, `"tanh"`, `"logistic"` |
| `solver` | Verfahren zum Verbessern | `"adam"`, `"sgd"`, `"lbfgs"` |
| `lernrate` | wie grosse Korrekturschritte | z.B. 0.001 bis 0.1 |
| `epochen` | höchstens so viele Runden | z.B. 500 |

```python
class Experiment:
    """Führt das ganze Experiment durch, Schritt für Schritt."""

    def __init__(self, daten, test_anteil=0.25, neuronen=(8,),
                 aktivierung="relu", solver="adam", lernrate=0.01, epochen=500):
        self.daten = daten
        self.test_anteil = test_anteil

        # Wir merken uns die Einstellungen, damit vergleiche() sie
        # später wiederverwenden kann.
        self.einstellungen = dict(hidden_layer_sizes=neuronen,
                                  activation=aktivierung,
                                  solver=solver,
                                  learning_rate_init=lernrate,
                                  max_iter=epochen,
                                  random_state=42)

        self.modell = MLPClassifier(**self.einstellungen)

    # --- die einzelnen Schritte ---------------------------------------

    def aufteilen(self):
        """Wir testen nie mit denselben Daten, mit denen wir gelernt haben.

        stratify sorgt dafür, dass in beiden Teilen gleich viel Prozent
        aktive Moleküle landen. Unsere Datei ist nämlich sortiert: zuerst
        alle aktiven, dann alle inaktiven.
        """
        self.X_train, self.X_test, self.y_train, self.y_test = train_test_split(
            self.daten.X, self.daten.y,
            test_size=self.test_anteil,
            random_state=42,        # immer dieselbe Aufteilung
            stratify=self.daten.y,
        )

    def skalieren(self):
        """Die Masse liegt bei ungefähr 500, die Anzahl Donoren bei 2.
        Ohne Umrechnung würde die Masse alles dominieren.

        Achtung: Der Skalierer lernt nur aus den Trainingsdaten. Sonst
        würde Wissen aus den Testdaten ins Training sickern.
        """
        self.skalierer = StandardScaler()

        self.X_train = self.skalierer.fit_transform(self.X_train)  # lernen und anwenden
        self.X_test = self.skalierer.transform(self.X_test)        # nur anwenden

    def trainieren(self):
        """Das Netz geht die Daten viele Male durch und wird dabei besser."""
        self.modell.fit(self.X_train, self.y_train)

        verlauf = self.modell.loss_curve_
        print(f"Fertig nach {len(verlauf)} Epochen")
        print(f"Verlust am Anfang: {verlauf[0]:.3f}")
        print(f"Verlust am Ende:   {verlauf[-1]:.3f}")

    # --- bewerten ------------------------------------------------------

    def genauigkeit(self, X, y):
        """Anteil der richtigen Antworten."""
        return (self.modell.predict(X) == y).mean()

    def verlust(self, X, y):
        """Wie falsch liegt das Modell? Kleiner ist besser."""
        p = self.modell.predict_proba(X)[:, 1]   # Wahrscheinlichkeit für "aktiv"
        p = np.clip(p, 1e-7, 1 - 1e-7)           # nie genau 0 oder 1, wegen log
        return -np.mean(y * np.log(p) + (1 - y) * np.log(1 - p))

    def auswerten(self):
        print("                  Training    Test")
        print(f"Genauigkeit:         {self.genauigkeit(self.X_train, self.y_train):.3f}   "
              f"{self.genauigkeit(self.X_test, self.y_test):.3f}")
        print(f"Verlust:             {self.verlust(self.X_train, self.y_train):.3f}   "
              f"{self.verlust(self.X_test, self.y_test):.3f}")

        raten = max(self.y_test.mean(), 1 - self.y_test.mean())
        print(f"\nWer immer die häufigere Klasse rät, kommt auf {raten:.3f}")

    def plot_verlust(self):
        """Die Kurve sollte fallen: das Netz macht weniger Fehler."""
        plt.plot(self.modell.loss_curve_)
        plt.xlabel("Epoche")
        plt.ylabel("Verlust")
        plt.title("Das Netz wird besser")
        plt.grid(alpha=0.3)
        plt.show()

    def konfusionsmatrix(self):
        """Wo genau liegt das Modell falsch?"""
        vorhersage = self.modell.predict(self.X_test)

        print("                   vorhergesagt")
        print("                 inaktiv   aktiv")
        for echt, bezeichnung in [(0, "echt inaktiv"), (1, "echt aktiv  ")]:
            zeile = [np.sum((self.y_test == echt) & (vorhersage == p)) for p in (0, 1)]
            print(f"{bezeichnung}  {zeile[0]:>7}{zeile[1]:>8}")

    # --- benutzen ------------------------------------------------------

    def vorhersagen(self, mw, alogp, hba, hbd, rb, name="Neues Molekül"):
        """Ein einzelnes Molekül eingeben und beurteilen lassen."""
        neu = np.array([[mw, alogp, hba, hbd, rb]])
        neu = self.skalierer.transform(neu)    # gleich umrechnen wie beim Training

        sicherheit = self.modell.predict_proba(neu)[0][1]
        antwort = "wirkt vermutlich" if sicherheit > 0.5 else "wirkt vermutlich nicht"

        print(f"{name:<18} {antwort:<24} (Sicherheit: {sicherheit:.0%})")
        return sicherheit

    def run(self):
        """Alle Schritte in der richtigen Reihenfolge."""
        self.aufteilen()
        self.skalieren()
        self.trainieren()
        self.auswerten()

    # --- mehrere Einstellungen ausprobieren -----------------------------

    def vergleiche(self, varianten):
        """Probiert mehrere Einstellungen aus und zeigt eine Tabelle.

        varianten ist eine Liste von Dictionaries, zum Beispiel:
            [{"neuronen": (8,)},
             {"neuronen": (32, 16), "aktivierung": "tanh"}]

        Alles, was nicht genannt wird, bleibt wie im Hauptexperiment.
        """
        print(f"{'Neuronen':<14}{'Aktivierung':<13}{'Solver':<8}"
              f"{'Verlust':>9}{'Training':>10}{'Test':>8}")
        print("-" * 62)

        for v in varianten:
            e = dict(self.einstellungen)       # Kopie der Grundeinstellungen
            e["hidden_layer_sizes"] = v.get("neuronen", e["hidden_layer_sizes"])
            e["activation"] = v.get("aktivierung", e["activation"])
            e["solver"] = v.get("solver", e["solver"])
            e["learning_rate_init"] = v.get("lernrate", e["learning_rate_init"])
            e["max_iter"] = v.get("epochen", e["max_iter"])

            m = MLPClassifier(**e).fit(self.X_train, self.y_train)

            train = (m.predict(self.X_train) == self.y_train).mean()
            test = (m.predict(self.X_test) == self.y_test).mean()

            print(f"{str(e['hidden_layer_sizes']):<14}{e['activation']:<13}"
                  f"{e['solver']:<8}{m.loss_:>9.3f}{train:>10.3f}{test:>8.3f}")
```

## Teil 3: Schritt für Schritt durchgehen

Statt gleich `run()` aufzurufen, machen wir die Schritte einzeln, damit man
sieht, was jeder bewirkt.

Zuerst das Objekt erzeugen. Hier dürft ihr alles verändern.

```python
experiment = Experiment(daten,
                        neuronen=(8,),        # eine Schicht mit 8 Neuronen
                        aktivierung="relu",   # relu, tanh, logistic
                        solver="adam",        # adam, sgd, lbfgs
                        lernrate=0.01,
                        epochen=500)

print("Experiment bereit")
```

### Schritt 1: Aufteilen

```python
experiment.aufteilen()

print(f"Training: {len(experiment.y_train)} Moleküle, "
      f"davon {experiment.y_train.mean():.1%} aktiv")
print(f"Test:     {len(experiment.y_test)} Moleküle, "
      f"davon {experiment.y_test.mean():.1%} aktiv")
```

Beide Teile haben denselben Anteil aktiver Moleküle. Das ist `stratify`.

### Schritt 2: Standardisieren

Jede Spalte wird so verschoben und gestaucht, dass sie Mittelwert 0 und
Standardabweichung 1 hat:

$$ x_{neu} = \frac{x - \text{Mittelwert}}{\text{Standardabweichung}} $$

```python
print("Vorher, erstes Trainingsmolekül:")
print(experiment.X_train[0].round(2))

experiment.skalieren()

print()
print("Nachher:")
print(experiment.X_train[0].round(2))
```

Zur Kontrolle: Die Trainingsdaten haben jetzt pro Spalte Mittelwert 0 und
Streuung 1. Die Testdaten nicht ganz, und das ist richtig so: sie wurden mit
den Werten der Trainingsdaten umgerechnet.

```python
print("Training, Mittelwerte:", experiment.X_train.mean(axis=0).round(3))
print("Training, Streuungen: ", experiment.X_train.std(axis=0).round(3))
print()
print("Test, Mittelwerte:    ", experiment.X_test.mean(axis=0).round(3))
```

### Schritt 3: Trainieren

So lernt das Netz:

1. Es rät zuerst zufällig.
2. Eine **Verlustfunktion** misst, wie falsch es lag.
3. Es korrigiert seine internen Zahlen ein wenig.
4. Zurück zu Schritt 1. Eine solche Runde heisst **Epoche**.

```python
experiment.trainieren()
```

Das Netz hat von selbst aufgehört, bevor die 500 Epochen um waren. Es merkt,
wenn sich nichts mehr verbessert.

```python
experiment.plot_verlust()
```

### Was misst die Verlustfunktion eigentlich?

Sie bestraft falsche Vorhersagen, und zwar umso härter, je selbstsicherer das
Modell daneben lag. Das ist die einzige Formel im ganzen Notebook, und man
sieht sie am besten an Zahlen:

```python
for p in [0.99, 0.9, 0.6, 0.5, 0.4, 0.1, 0.01]:
    print(f"Modell sagt {p:>5.0%} aktiv, Molekül ist aktiv  -> Kosten {-np.log(p):.2f}")
```

Sicher und richtig kostet fast nichts, sicher und falsch kostet sehr viel.
Deshalb lernt das Modell, vorsichtig zu sein, wenn es sich nicht sicher ist.

### Schritt 4: Auswerten

Die Genauigkeit allein sagt wenig. Erst der Vergleich mit blossem Raten zeigt,
ob das Modell überhaupt etwas gelernt hat.

```python
experiment.auswerten()
```

Und wo genau liegt es falsch?

```python
experiment.konfusionsmatrix()
```

## Teil 4: Ein eigenes Molekül testen

Das neue Molekül muss **genau gleich umgerechnet** werden wie die
Trainingsdaten. Deshalb benutzt `vorhersagen()` denselben Skalierer.

```python
#                      MW     AlogP  HBA HBD RB
experiment.vorhersagen(180.2,   1.31,  3,  1,  2, name="Aspirin")
experiment.vorhersagen(194.2,  -1.03,  3,  0,  0, name="Koffein")
experiment.vorhersagen(206.3,   3.07,  1,  1,  4, name="Ibuprofen")
experiment.vorhersagen(431.6,   5.04,  4,  2,  5, name="BACE-Inhibitor")
```

**Schaut euch das Ergebnis kritisch an.** Koffein ist kein Wirkstoff gegen
Alzheimer, und das letzte Molekül stammt aus unserem Datensatz und wirkt
wirklich. Das Modell liegt hier also daneben.

Der Grund: Es kennt von jedem Molekül nur fünf Zahlen. Aspirin, Koffein und
Ibuprofen sind ausserdem viel kleiner als alles, was es im Training gesehen
hat. Ein Modell ausserhalb seines gewohnten Bereichs rät im Grunde.

Das ist keine Panne, sondern die wichtigste Lektion des Tages: **Ein Modell
ist nur so gut wie die Daten, die es bekommt.**

## Teil 5: Einstellungen vergleichen

Welche Netzstruktur ist die beste? Das probiert man aus.

```python
experiment.vergleiche([
    {"neuronen": (2,)},
    {"neuronen": (8,)},
    {"neuronen": (32, 16)},
    {"neuronen": (32, 16), "aktivierung": "tanh"},
    {"neuronen": (32, 16), "aktivierung": "logistic", "solver": "sgd"},
])
```

Schaut euch die Zeile mit `tanh` genau an: bester Verlust, beste Genauigkeit
auf den Trainingsdaten, und trotzdem nicht das beste Modell auf den Testdaten.
Es hat die Trainingsdaten auswendig gelernt, statt die Regel dahinter zu
verstehen. Das nennt man **Überanpassung**, und es ist der häufigste Fehler
im Machine Learning.

Die letzte Zeile zeigt das Gegenteil: Mit `logistic` und `sgd` lernt das Netz
kaum etwas.
